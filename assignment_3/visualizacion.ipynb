```python
import pandas as pd
import plotly.express as px

data = pd.read_csv("datos/tabla-final.csv")
data
```

Indicar qué tema o pregunta se abordará y qué columnas se usarán.

- Estamos abordando la siguiente pregunta: ¿Los departamentos con más lluvia son también aquellos con mayor cantidad de días de lluvia fuerte? Se busca entender si llover mucho equivale a tener lluvias intensas. 

- Se usarán las columnas "departamento", "lluvia_total_mm" y "dias_lluvia_fuerte"

```python
# Gráfico de barras

top = data.sort_values("lluvia_total_mm", ascending=False).head(15)

fig_bar = px.bar(top, x="departamento", y="lluvia_total_mm",
             title="Top 10 departamentos con más lluvia (mm)<br><sup>Año 2022</sup>",
             labels={"departamento": "Departamentos", "lluvia_total_mm": "Lluvia total (mm)"},
             text="lluvia_total_mm",
             height=750)
fig_bar.update_traces(texttemplate="%{text:,.0f}", textposition="outside", cliponaxis=False)
fig_bar.update_layout(title_x=0.5)
fig_bar.show()
```

```python
# Gráfico de dispersión

fig_scatter = px.scatter(data, x="lluvia_total_mm", y="dias_lluvia_fuerte",
                 hover_name="departamento",
                 title="Lluvia total vs. días de lluvia fuerte por departamento<br><sup>Año 2022</sup>",
                 labels={"lluvia_total_mm": "Lluvia total (mm)",
                         "dias_lluvia_fuerte": "Días de lluvia fuerte"})
fig_scatter.update_layout(title_x=0.5)
fig_scatter.show()
```

```python
# Gráfico libre - tree map

fig_treemap = px.treemap(data, path=[px.Constant("Lluvia total en mm"), "departamento"],
                 values="lluvia_total_mm",
                 color="dias_lluvia_fuerte",
                 color_continuous_scale=["#91bfdb", "#ffffbf", "#fc8d59"],
                 title="Lluvia total y días de lluvia fuerte por departamento<br><sup>Año 2022</sup>",
                 labels={"lluvia_total_mm": "Lluvia total (mm)",
                         "dias_lluvia_fuerte": "Días de lluvia fuerte"})
fig_treemap.update_layout(title_x=0.5)
fig_treemap.update_traces(
    pathbar_visible=False,
    texttemplate="<b>%{label}</b><br>%{value} mm",
    hovertemplate="<b>%{label}</b><br>Lluvia total: %{value} mm"
                  "<br>Días de lluvia fuerte: %{color}<extra></extra>"
)
fig_treemap.show()
```

```python
# Verificación rápida 

cols = ["departamento", "lluvia_total_mm", "dias_lluvia_fuerte"]

# Top 3 departamentos con más lluvia total

print("Top 3 con más lluvia total:")
print(data.nlargest(3, "lluvia_total_mm")[cols], "\n")
```

Coincide con el gráfico de barras y con los rectangulos más grandes del tree map. 

```python
# Top 3 departamentos con más días de lluvia fuerte

print("Top 3 con más días de lluvia fuerte:")
print(data.nlargest(3, "dias_lluvia_fuerte")[cols], "\n")
```

Coincide con los valores más altos del gráfico de dispersión

```python
# Top departamentos con menos lluvia total 

print("Top 3 con menos lluvia total:")
print(data.nsmallest(3, "lluvia_total_mm")[cols], "\n")
```

Coincide con los rectángulos más pequeños del tree map.

```python
# Exportar un gráfico a html

fig_treemap.write_html("resultados/grafico.html", include_plotlyjs="cdn")
print("Treemap exportado en resultados/grafico.html")
```
